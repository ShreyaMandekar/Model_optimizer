```python
import torch
import torch.nn as nn

class TinyTransformerBlock(nn.Module):
    def __init__(self, d_model=128, nhead=2, ff_dim=256):
        super().__init__()
        self.attn    = nn.MultiheadAttention(d_model, nhead, batch_first=True)
        self.norm1   = nn.LayerNorm(d_model)
        self.norm2   = nn.LayerNorm(d_model)
        self.ff      = nn.Sequential(
            nn.Linear(d_model, ff_dim),
            nn.GELU(),
            nn.Linear(ff_dim, d_model),
        )

    def forward(self, x):
        # self attention + residual
        attn_out, _ = self.attn(x, x, x)
        x = self.norm1(x + attn_out)
        # feedforward + residual
        x = self.norm2(x + self.ff(x))
        return x


class TinyTransformer(nn.Module):
    """
    Small transformer for classification.
    Designed to trace cleanly with torch.fx.
    Comparable size to bert-tiny (~4M params).
    """
    def __init__(
        self,
        vocab_size=1000,
        d_model=128,
        nhead=2,
        ff_dim=256,
        num_layers=2,
        num_classes=2,
        max_seq_len=128,
    ):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, d_model)
        # fixed positional embedding table — no dynamic arange, traces cleanly
        self.pos_embedding = nn.Embedding(max_seq_len, d_model)

        self.layers = nn.ModuleList([
            TinyTransformerBlock(d_model, nhead, ff_dim)
            for _ in range(num_layers)
        ])
        self.classifier = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, num_classes),
        )

    def forward(self, input_ids, position_ids):
        # position_ids passed explicitly — no dynamic arange inside forward
        x = self.embedding(input_ids) + self.pos_embedding(position_ids)
        for layer in self.layers:
            x = layer(x)
        # classify using [CLS] token (first token)
        x = self.classifier(x[:, 0, :])
        return x


# build model
fp32_model = TinyTransformer(
    vocab_size=VOCAB_SIZE,
    d_model=128,
    nhead=2,
    ff_dim=256,
    num_layers=2,
    num_classes=2,
).to(DEVICE).eval()

total_params = sum(p.numel() for p in fp32_model.parameters())
print(f"Parameters: {total_params:,}")
print(f"FP32 model ready ✓")
```

```python
from torch.ao.quantization import get_default_qat_qconfig_mapping, QConfigMapping
from torch.ao.quantization.quantize_fx import prepare_qat_fx

# fresh CPU copy
m = TinyTransformer(
    vocab_size=VOCAB_SIZE,
    d_model=128,
    nhead=2,
    ff_dim=256,
    num_layers=2,
    num_classes=2,
).eval()

# default qconfig but exclude embedding layers
# embedding quantization needs special float_qparams scheme
# and is not relevant to our fusion research
qconfig_mapping = get_default_qat_qconfig_mapping("x86")
qconfig_mapping.set_module_name("embedding", None)
qconfig_mapping.set_module_name("pos_embedding", None)

example_inputs = (
    torch.randint(0, VOCAB_SIZE, (BATCH_SIZE, SEQ_LEN)),
    torch.arange(SEQ_LEN).unsqueeze(0).expand(BATCH_SIZE, -1),
)

qat_model = prepare_qat_fx(m, qconfig_mapping, example_inputs)
qat_model  = qat_model.to(DEVICE).eval()

# count FakeQuantize nodes
# correct detection for PyTorch 2.5
fq_nodes = [
    n for n in qat_model.graph.nodes
    if "activation_post_process" in str(n.target).lower()
    or "fake_quant" in str(n.target).lower()
]
print(f"FakeQuantize nodes inserted: {len(fq_nodes)}")
print("QAT model ready ✓")
```

```python
# shows exactly where FakeQuantize nodes landed
# these are your fusion break locations

fq_names = {n.name for n in fq_nodes}

print(f"{'OP':<16} {'TARGET':<50} NOTE")
print("-"*90)

for node in qat_model.graph.nodes:
    if node.op in ("placeholder", "output"):
        continue
    is_fq  = (node.name in fq_names or "fake_quant" in str(node.target).lower())
    marker = "  ◄── FakeQuantize" if is_fq else ""
    print(f"{node.op:<16} {str(node.target)[:50]:<50}{marker}")
```

```python
# Task 2.1 — Empirical fusion discovery
# For each op pair, measure kernel count with and without a QDQ between them
# If kernel count increases when QDQ is inserted → that pair fuses → QDQ breaks it

import torch
import torch.nn as nn
from torch.profiler import profile, ProfilerActivity

def count_kernels_simple(model, x):
    """Count unique CUDA kernels fired by a model."""
    model.eval()
    # warmup
    with torch.no_grad():
        for _ in range(3):
            model(x)
    torch.cuda.synchronize()

    with profile(activities=[ProfilerActivity.CUDA]) as prof:
        with torch.no_grad():
            model(x)
    torch.cuda.synchronize()

    return len([e for e in prof.key_averages() if e.device_time_total > 0])


# ── define op pairs to test ──────────────────────────────────────
# each entry: (name, model_without_qdq, model_with_qdq)

x = torch.randn(4, 64).cuda()

pairs_to_test = {

    "linear+relu": (
        nn.Sequential(nn.Linear(64, 64), nn.ReLU()).cuda(),
        nn.Sequential(nn.Linear(64, 64), nn.Flatten(0), nn.Unflatten(0, (4,)),
                      nn.ReLU()).cuda(),  # placeholder — we insert QDQ manually below
    ),
}

# ── cleaner approach: test with torch.compile ─────────────────────

def make_fused(op_a, op_b):
    """Two ops with no QDQ — compiler can fuse."""
    class Fused(nn.Module):
        def forward(self, x):
            return op_b(op_a(x))
    return Fused().cuda()

def make_unfused(op_a, op_b):
    """Two ops with explicit intermediate materialization — simulates QDQ break."""
    class Unfused(nn.Module):
        def forward(self, x):
            x = op_a(x)
            x = x.clone()   # clone forces materialization, like QDQ does
            return op_b(x)
    return Unfused().cuda()


# ── test each pair ────────────────────────────────────────────────

d = 128  # match your TinyTransformer's d_model

op_pairs = [
    ("linear → relu",      nn.Linear(d, d).cuda(), nn.ReLU().cuda()),
    ("linear → gelu",      nn.Linear(d, d).cuda(), nn.GELU().cuda()),
    ("linear → layernorm", nn.Linear(d, d).cuda(), nn.LayerNorm(d).cuda()),
    ("relu → add",         nn.ReLU().cuda(),        None),   # add is special, tested separately
    ("layernorm → linear", nn.LayerNorm(d).cuda(),  nn.Linear(d, d).cuda()),
    ("gelu → linear",      nn.GELU().cuda(),        nn.Linear(d, d).cuda()),
]

x = torch.randn(4, d).cuda()

print(f"{'Op Pair':<25} {'Fused kernels':>15} {'Unfused kernels':>17} {'Breaks fusion?':>15}")
print("-"*75)

results = {}
for name, op_a, op_b in op_pairs:
    if op_b is None:
        continue

    fused_model   = make_fused(op_a, op_b)
    unfused_model = make_unfused(op_a, op_b)

    k_fused   = count_kernels_simple(fused_model,   x)
    k_unfused = count_kernels_simple(unfused_model, x)

    breaks_fusion = k_unfused > k_fused
    results[name] = {
        "fused_kernels":   k_fused,
        "unfused_kernels": k_unfused,
        "breaks_fusion":   breaks_fusion,
    }

    print(f"{name:<25} {k_fused:>15} {k_unfused:>17} {str(breaks_fusion):>15}")

print("-"*75)
print("\nThese results tell us which QDQ placements actually matter.")
print("Only penalize QDQ nodes at pairs where breaks_fusion=True.")
```
